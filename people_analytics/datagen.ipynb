# Goal of generating a new synthetic dataset

```python
# Dataset parameters: 
# Number of total samples: 4500

# The dataset is generated using the following code:

import random
import pandas as pd

random.seed(0)

# columns: EmployeeID, EmployeeLEvel, TenureYears, ProductGroup, Country, WorkLocation, TeamSize, PerformanceRating, SatisfactionScore


# step 1, generate EmployeeID
# EmployeeID: randomly generated six-digit number
employee_ids = [random.randint(100000, 999999) for _ in range(4500)]

# step 2, generate EmployeeLevel
# EmployeeLevel: 1-4, with lots of 1s and 2s, and fewer 3s, and fewer 4s
employee_levels = [random.choices([1, 2, 3, 4], weights=[0.5, 0.3, 0.15, 0.05])[0] for _ in range(4500)]

# step 3, generate TenureYears
# EmployeeLevel 1 has tenure mostly in years 1-4 with few 5-8, EmployeeLevel 2 has mostly tenure in years 1-10 with few 11-14
def generate_tenure_years(level):
    if level == 1:
        return random.choices(range(1, 9), weights=[0.3, 0.3, 0.2, 0.1, 0.05, 0.03, 0.02, 0.01])[0]
    elif level == 2:
        return random.choices(range(1, 15), weights=[0.2, 0.2, 0.15, 0.1, 0.1, 0.07, 0.05, 0.04, 0.03, 0.02, 0.01, 0.01, 0.005, 0.005])[0]
    elif level == 3:
        return random.choices(range(1, 20), weights=[0.15, 0.15, 0.1, 0.1, 0.08, 0.07, 0.06, 0.05, 0.04, 0.03, 0.02, 0.02, 0.01, 0.01, 0.01, 0.005, 0.005, 0.005, 0.005])[0]
    elif level == 4:
        return random.choices(range(1, 25), weights=[0.1, 0.1, 0.1, 0.08, 0.08, 0.07, 0.06, 0.05, 0.04, 0.03, 0.03, 0.02, 0.02, 0.01, 0.01, 0.01, 0.01])[0]
    
tenure_years = [generate_tenure_years(level) for level in employee_levels]

# step 4, generate ProductGroup
# ProductGroup: either "GroupA" or "GroupB" or "GroupC" or "GroupD"
# The number of samples for each ProductGroup is as follows: GroupA: 2500, GroupB: 1000, GroupC: 800, GroupD: 200
product_groups = random.choices(["GroupA", "GroupB", "GroupC", "GroupD"], weights=[0.55, 0.25, 0.2, 0.05], k=4500)

# step 5, generate Country
# Country is either "United States" or "Canada" or "India"
# GroupA has people in all three countries, GroupB has people only in the United States and Canada, GroupC has people only in the United States and India, and GroupD has people only in the United States
def generate_country(group):
    if group == "GroupA":
        return random.choices(["United States", "Canada", "India"], weights=[0.4, 0.3, 0.3])[0]
    elif group == "GroupB":
        return random.choices(["United States", "Canada"], weights=[0.7, 0.3])[0]
    elif group == "GroupC":
        return random.choices(["United States", "India"], weights=[0.6, 0.4])[0]
    elif group == "GroupD":
        return "United States"

# step 6, generate WorkLocation
# WorkLocation: either "onsite", "hybrid", or "remote"
# GroupA has people in a mix of all three work locations, GroupB has people only onsite, GroupC has people in a mix of all three, and GroupD has people only in hybrid or remote
def generate_work_location(group):
    if group == "GroupA":
        return random.choices(["onsite", "hybrid", "remote"], weights=[0.4, 0.3, 0.3])[0]
    elif group == "GroupB":
        return "onsite"
    elif group == "GroupC":
        return random.choices(["onsite", "hybrid", "remote"], weights=[0.3, 0.3, 0.4])[0]
    elif group == "GroupD":
        return random.choices(["hybrid", "remote"], weights=[0.6, 0.4])[0]
    
work_locations = [generate_work_location(group) for group in product_groups]

# step 7, generate TeamSize
# TeamSize: 1-20, with most being on teams of 5-12, and fewer on teams of 1-4, and even fewer on teams of 13-20
# GroupD is mostly smaller team sizes, GroupC is mostly larger team sizes
def generate_team_size(group):
    if group == "GroupA":
        return random.choices(range(1, 21), weights=[0.05, 0.1, 0.15, 0.2, 0.2, 0.15, 0.1, 0.05, 0.03, 0.02, 0.02, 0.01, 0.01])[0]
    elif group == "GroupB":
        return random.choices(range(1, 21), weights=[0.1, 0.15, 0.2, 0.25, 0.15, 0.1, 0.05, 0.03, 0.02, 0.01, 0.01])[0]
    elif group == "GroupC":
        return random.choices(range(1, 21), weights=[0.01, 0.02, 0.03, 0.05, 0.1, 0.15, 0.2, 0.25, 0.15, 0.1, 0.05, 0.03, 0.02, 0.01, 0.01])[0] 
    elif group == "GroupD":
        return random.choices(range(1, 21), weights=[0.2, 0.25, 0.3, 0.15, 0.1, 0.05, 0.03, 0.02])[0]
    
team_sizes = [generate_team_size(group) for group in product_groups]

# step 8, generate PerformanceRating
# PerformanceRating: 1-5, with lots of 3s and 4s, and fewer 1s, 2s, and 5s
# Employees in GroupC are generally lower performers, GroupD are generally higher performers
def generate_performance_rating(group):
    if group == "GroupA":
        return random.choices(range(1, 6), weights=[0.1, 0.15, 0.2, 0.3, 0.25])[0]
    elif group == "GroupB":
        return random.choices(range(1, 6), weights=[0.1, 0.15, 0.3, 0.3, 0.15])[0]
    elif group == "GroupC":
        return random.choices(range(1, 6), weights=[0.3, 0.25, 0.2, 0.15, 0.1])[0]
    elif group == "GroupD":
        return random.choices(range(1, 6), weights=[0.05, 0.1, 0.2, 0.3, 0.35])[0]
    
performance_ratings = [generate_performance_rating(group) for group in product_groups]

# step 9, generate SatisfactionScore
# SatisfactionScore: 1-5, with lots of 3s and 4s, and fewer 1s, 2s, and 5s
# People in GroupC are generally unhappy, and people in GroupD are generally happy

def generate_satisfaction_score(group):
    if group == "GroupA":
        return random.choices(range(1, 6), weights=[0.1, 0.15, 0.2, 0.3, 0.25])[0]
    elif group == "GroupB":
        return random.choices(range(1, 6), weights=[0.1, 0.15, 0.3, 0.3, 0.15])[0]
    elif group == "GroupC":
        return random.choices(range(1, 6), weights=[0.3, 0.25, 0.2, 0.15, 0.1])[0]
    elif group == "GroupD":
        return random.choices(range(1, 6), weights=[0.05, 0.1, 0.2, 0.3, 0.35])[0]
    
satisfaction_scores = [generate_satisfaction_score(group) for group in product_groups]

# create a DataFrame
data = {
    'EmployeeID': employee_ids,
    'EmployeeLevel': employee_levels,
    'TenureYears': tenure_years,
    'ProductGroup': product_groups,
    'Country': [generate_country(group) for group in product_groups],
    'WorkLocation': work_locations,
    'TeamSize': team_sizes,
    'PerformanceRating': performance_ratings,
    'SatisfactionScore': satisfaction_scores
}

df = pd.DataFrame(data)

# save the DataFrame to a CSV file
df.to_csv('employee_data.csv', index=False)

print("Data generation complete.")
```
