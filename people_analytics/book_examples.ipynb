```python
# load libraries
import pandas as pd 
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
import seaborn as sns
import altair as alt
```

```python
# get data
url = "http://peopleanalytics-regression-book.org/data/ugtests.csv"
ugtests = pd.read_csv(url)
```

```python
# summary statistics, round to 1 decimal
print(ugtests.describe().round(1))
```

```python
# print a pairplot
sns.pairplot(ugtests, corner = True)
```

```python
# get the correlation coefficients and round to 2 decimal places
correlation = ugtests.corr().round(2)
print(correlation)
```

```python
model1 = smf.ols(formula = "Final ~ Yr3 + Yr2 + Yr1", data = ugtests) # define model
ugtests_model1 = model1.fit() # fit model
print(ugtests_model1.summary()) # print results
```

```python
model2 = smf.ols(formula = "Final ~ Yr3 + Yr2", data = ugtests) # define model
ugtests_model2 = model2.fit() # fit model
print(ugtests_model2.summary()) # print results
```
